```python
from sentence_transformers import SentenceTransformer
from sklearn.feature_extraction.text import TfidfTransformer
```

```python
print("="*50)
print("Loading Sentence Transformer...")
print("="*50)
encoder = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
```

```python
def encoding(row):
    return encoder.encode(f"{row['Ticket_Subject']} | {row['Ticket_Description']}")
```

```python
import pandas as pd
```

```python
df = pd.read_csv("../dataset/preprocessed_data.csv")
```

```python
results = []
for idx,row in df.iterrows():
    if(idx+1)%1000==0:
        print(f"{idx+1} Completed")
    results.append(encoding(row))
    
```

```python
encod_df = pd.DataFrame(results,index=df.Ticket_ID)
encod_df
```

```python
df = df.set_index("Ticket_ID")
```

```python
df
```

```python
df["Final_severity_label"] = df["Final_severity_label"].map({"LOW":0,"MEDIUM":1,"HIGH":2,"CRITICAL":3})
df["Priority_Level"] = df["Priority_Level"].map({"Low":0,"Medium":1,"High":2,"Critical":3})
```

```python
encod_df["Target"] = df["Final_severity_label"]
priority = df["Priority_Level"]
```

```python
encod_df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_Train,X_test,y_train,y_test = train_test_split(encod_df.drop(columns=["Target"]), encod_df["Target"], test_size=0.2 ,stratify=encod_df["Target"])
```

```python
priority_test = priority.loc[y_test.index]
```

```python
priority_train = priority.loc[y_train.index]
```

```python
mismatch_test = df["Is_Mismatch"].loc[y_test.index]
mismatch_train = df["Is_Mismatch"].loc[y_train.index]
```

```python
X_Train.shape
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
model = RandomForestClassifier()
model.fit(X_Train,y_train)
```

```python
from sklearn.metrics import classification_report
```

```python
y_pred = model.predict(X_test)
report = classification_report(y_test,y_pred)
print("Classfication report\n")
print(report)
print("Mismatch Report\n")
yp_mismatch = (priority_test != y_pred).astype(int)
report = classification_report(mismatch_test,yp_mismatch)
print(report)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler() 
X_train[['Resolution_Time_Hours','Satisfaction_Score']] = scaler.fit_transform(X_train[['Resolution_Time_Hours','Satisfaction_Score']])
X_test[['Resolution_Time_Hours','Satisfaction_Score']] = scaler.transform(X_test[['Resolution_Time_Hours','Satisfaction_Score']])   
```

```python
mismatch_test = df["Is_Mismatch"].loc[y_test.index]
mismatch_train = df["Is_Mismatch"].loc[y_train.index]
priority_train = df["Priority_Level"].loc[y_train.index]
priority_test = df["Priority_Level"].loc[y_test.index]
```

```python
X_train.columns
```

```python
emb_model = XGBClassifier()
meta_model = XGBClassifier()

emb_model.fit(X_train.drop(columns=['Resolution_Time_Hours','Satisfaction_Score','Ticket_Channel_Email','Ticket_Channel_Web Form','Issue_Category_Billing','Issue_Category_Fraud','Issue_Category_General Inquiry','Issue_Category_Technical']),y_train)
meta_model.fit(X_train[['Resolution_Time_Hours','Satisfaction_Score','Ticket_Channel_Email','Ticket_Channel_Web Form','Issue_Category_Billing','Issue_Category_Fraud','Issue_Category_General Inquiry','Issue_Category_Technical']],y_train)

emb_prob = emb_model.predict_proba(X_test.drop(columns=['Resolution_Time_Hours','Satisfaction_Score','Ticket_Channel_Email','Ticket_Channel_Web Form','Issue_Category_Billing','Issue_Category_Fraud','Issue_Category_General Inquiry','Issue_Category_Technical']))
meta_prob = meta_model.predict_proba(X_test[['Resolution_Time_Hours','Satisfaction_Score','Ticket_Channel_Email','Ticket_Channel_Web Form','Issue_Category_Billing','Issue_Category_Fraud','Issue_Category_General Inquiry','Issue_Category_Technical']])
```

```python
final_prob = 0.6*emb_prob + 0.4*meta_prob
```

```python
final_prob
```

```python
import numpy as np
```

```python
y_pred = []
for prob in final_prob:
    y_pred.append(np.argmax(prob))
```

```python
report = classification_report(y_test,y_pred)
print("Classfication report\n")
print(report)
print("Mismatch Report\n")
yp_mismatch = (priority_test != y_pred).astype(int)
report = classification_report(mismatch_test,yp_mismatch)
print(report)
```
