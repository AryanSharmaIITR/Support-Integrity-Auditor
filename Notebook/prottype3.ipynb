```python
from sentence_transformers import SentenceTransformer
from sklearn.feature_extraction.text import TfidfTransformer
```

```python
print("="*50)
print("Loading Sentence Transformer...")
print("="*50)
encoder = SentenceTransformer('sentence-transformers/all-MiniLM-L6-v2')
```

```python
def encoding(row):
    return encoder.encode(f"{row['Ticket_Subject']} | {row['Ticket_Description']} | {row['Issue_Category']} | {row['Ticket_Channel']} | {row['Submission_Date']} | {row['Resolution_Time_Hours']} | {row['Satisfaction_Score']}")
```

```python
import pandas as pd
```

```python
df = pd.read_csv("../dataset/preprocessed_data.csv")
```

```python
results = []
for idx,row in df.iterrows():
    if(idx+1)%100==0:
        print(f"{idx+1} Completed")
    results.append(encoding(row))
    
```

```python
encod_df = pd.DataFrame(results,index=df.Ticket_ID)
encod_df
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
df = df.set_index("Ticket_ID")
```

```python
encod_df["Target"] = df["Is_Mismatch"]
```

```python
encod_df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_Train,X_test,y_train,y_test = train_test_split(encod_df.drop(columns=["Target"]), encod_df["Target"], test_size=0.2 ,stratify=encod_df["Target"])
```

```python
X_Train.shape
```

```python
model = RandomForestClassifier()
model.fit(X_Train,y_train)
```

```python
from sklearn.metrics import classification_report
```

```python
y_pred = model.predict(X_test)
report = classification_report(y_test,y_pred)

print(report)
```

```python
from xgboost import XGBClassifier
```

```python
model = XGBClassifier()
model.fit(X_Train,y_train)
```

```python
y_pred = model.predict(X_test)
report = classification_report(y_test,y_pred)

print(report)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
```

```python
model.fit(X_Train,y_train)
```

```python
y_pred = model.predict(X_test)
report = classification_report(y_test,y_pred)

print(report)
```
